```python
import pandas as pd

# File path (adjust if needed)
file_path = r"C:\Users\dsabatus\Documents\Tianle\my_code\Hearing_County_PUF.xlsx"

# Load the sheet "CrudePrevalance"
df = pd.read_excel(file_path, sheet_name="CrudePrevalence")

# Filter rows where GroupID == "GRP_001" - representing all ages, all ethnicities, and both genders
filtered_df = df[df["GroupID"] == "GRP_001"]
```

```python
print(filtered_df)
```

```python
# Keep only the desired columns
final_df = filtered_df[["FIPS", "MildHL_crd_rate", "ModSev_crd_rate"]]
```

```python
print(final_df)
```

```python
# Load the sheet "Overlays"
df = pd.read_excel(file_path, sheet_name="Overlays")

# Select only the desired columns
selected_columns = [
    "CountyState", 
    "audiologists_per_10k", 
    "HighRiskOccupation", 
    "FIPS", 
    "Population", 
    "Demog_age_0_14", 
    "Demog_age_15_64", 
    "Demog_age_65plus", 
    "Demog_urbanicity"
]
new_df = df[selected_columns]
```

```python
print(new_df)
```

```python
# Merge on FIPS (inner join keeps only matching rows)
merged_df = pd.merge(final_df, new_df, on="FIPS", how="inner")

print(f"Merged dataframe shape: {merged_df.shape}")
```

```python
print(merged_df)
```

```python
print(final_df["FIPS"].duplicated().sum(), "duplicates in final_df")
print(new_df["FIPS"].duplicated().sum(), "duplicates in new_df")
```

```python
# Find duplicated FIPS in merged_df
duplicates = merged_df[merged_df.duplicated(subset="FIPS", keep=False)]

print(f"Number of duplicated FIPS rows: {duplicates.shape[0]}")

# Sort and display duplicates
duplicates_sorted = duplicates.sort_values("FIPS")
print(duplicates_sorted)

# Count how many times each FIPS appears
duplicate_counts = duplicates_sorted["FIPS"].value_counts()
print("\nCount of each duplicated FIPS:")
print(duplicate_counts)
```

```python
# Remove duplicates based on FIPS, keeping the first occurrence
merged_unique = merged_df.drop_duplicates(subset="FIPS", keep="first")

print(f"Shape after removing duplicates: {merged_unique.shape}")
```

```python
print(merged_unique)
```

```python
output_path_csv = r"C:\Users\dsabatus\Documents\Tianle\my_code\Merged_Hearing_Data_unique.csv"
merged_unique.to_csv(output_path_csv, index=False)

print(f"Filtered and deduplicated file saved as CSV: {output_path_csv}")
```
